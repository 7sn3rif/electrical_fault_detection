# **Data Visualization**

```python
# G C B A are binary columns they used to detect faults so _colomuns=df.select_dtypes
output_columns = df.select_dtypes(include=['int64']).columns
output_columns
```

```python
# 3. Define the fault mapping dictionary
fault_mapping = {
    (0, 0, 0, 0): 'No Fault',
    (1, 0, 0, 1): 'LG Fault (Phase A-Gnd)',
    (0, 0, 1, 1): 'LL Fault (Phase A-B)',
    (1, 0, 1, 1): 'LLG Fault (Phase A-B-Gnd)',
    (0, 1, 1, 1): 'LLL Fault (Phases A-B-C)',
    (1, 1, 1, 1): 'LLLG Fault (Symmetrical)'
}

# 4. Zip the selected output columns into tuples and map to names
df['Fault_Tuple'] = list(zip(*[df[col] for col in output_columns]))
df['Fault_Name'] = df['Fault_Tuple'].map(fault_mapping).fillna('Other Fault')

# 5. Create Seaborn Bar Plot
sns.set_theme(style="whitegrid")
plt.figure(figsize=(10, 6))

ax = sns.countplot(
    data=df,
    x='Fault_Name',
    order=df['Fault_Name'].value_counts().index,
    palette='viridis'
)

# Customize title and labels
plt.title('Distribution of Electrical Fault Types', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Fault Category', fontsize=12, fontweight='bold')
plt.ylabel('Count', fontsize=12, fontweight='bold')
plt.xticks(rotation=30, ha='right')

# Add bar counts
for p in ax.patches:
    ax.annotate(
        f'{int(p.get_height())}',
        (p.get_x() + p.get_width() / 2., p.get_height()),
        ha='center', va='bottom', fontsize=10, xytext=(0, 3),
        textcoords='offset points'
    )

plt.tight_layout()
plt.show()
```

```python
# 2. Binary flag: 0 if all targets are 0, else 1
df['Binary_Fault'] = (df['G'] | df['C'] | df['B'] | df['A']).map({0: 'No Fault', 1: 'Fault'})

# 3. Create the pie chart
counts = df['Binary_Fault'].value_counts()

plt.figure(figsize=(6, 6))
plt.pie(
    counts,
    labels=counts.index,
    autopct='%1.1f%%',
    explode=[0, 0.08],
    colors=['#0c06c7', '#05daed'],
    startangle=180
)
plt.title('Fault vs No Fault Distribution', fontweight='bold')
plt.show()
```

```python


fig = px.scatter_3d(
    df,
    x='Ia', y='Ib', z='Ic',
    color='Fault_Name',
    title='3D Phase Current Space (Ia vs Ib vs Ic)',
    opacity=0.7,
    size_max=5
)
fig.update_traces(marker=dict(size=3))
fig.show()
```

```python
# Melt current columns into long form for easy multi-line plotting
df_currents = df.reset_index().melt(
    id_vars=['index', 'Fault_Name'],
    value_vars=['Ia', 'Ib', 'Ic'],
    var_name='Phase_Current',
    value_name='Amperes'
)

fig = px.line(
    df_currents,
    x='index',
    y='Amperes',
    color='Phase_Current',
    title='3-Phase Current Waveforms (Ia, Ib, Ic)',
    labels={'index': 'Sample Index'}
)
fig.show()
```

```python

#  Restructure data into Phase A, Phase B, and Phase C pairs
phase_a = df[['Va', 'Ia', 'Fault_Name']].rename(columns={'Va': 'Voltage', 'Ia': 'Current'}).assign(Phase='Phase A')
phase_b = df[['Vb', 'Ib', 'Fault_Name']].rename(columns={'Vb': 'Voltage', 'Ib': 'Current'}).assign(Phase='Phase B')
phase_c = df[['Vc', 'Ic', 'Fault_Name']].rename(columns={'Vc': 'Voltage', 'Ic': 'Current'}).assign(Phase='Phase C')

df_v_vs_i = pd.concat([phase_a, phase_b, phase_c], ignore_index=True)

#  Plot V vs I separated into subplots per line/phase
fig = px.scatter(
    df_v_vs_i,
    x='Voltage',
    y='Current',
    color='Fault_Name',
    facet_col='Phase',
    title='Voltage vs Current (V vs I) per Phase Line',
    labels={'Voltage': 'Voltage (V)', 'Current': 'Current (A)'},
    opacity=0.6,
    height=500
)

fig.update_layout(legend_title_text='Fault Type')
fig.show()
```
